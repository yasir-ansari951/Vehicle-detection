# **12. BUILDING IMAGE/XML PAIRS**

```python
print(" BUILDING IMAGE/XML PAIRS...")
print("=" * 60)

image_label_pairs = []

weather_types = ["Fog", "Rain", "Snow", "Sand"]

for weather in weather_types:

    weather_path = os.path.join(INPUT_PATH, weather, weather)

    if not os.path.exists(weather_path):
        continue

    # Find ALL xml files
    xml_files = []

    # Find ALL image files
    image_files = {}

    for root, dirs, files in os.walk(weather_path):

        for file in files:

            full_path = os.path.join(root, file)

            # XML
            if file.endswith(".xml"):
                xml_files.append(full_path)

            # Images
            elif file.endswith((".jpg", ".jpeg", ".png")):

                base = os.path.splitext(file)[0]

                image_files[base] = full_path

    # Match XML ↔ image
    for xml_path in xml_files:

        base = os.path.splitext(os.path.basename(xml_path))[0]

        if base in image_files:

            image_label_pairs.append(
                (image_files[base], xml_path)
            )

print(f"\n\n TOTAL VALID PAIRS: {len(image_label_pairs)}")

print("\n\n SAMPLE PAIRS:")
for pair in image_label_pairs[:5]:
    print(pair)
```

# **13. AUTO-DETECTING ALL CLASSES FROM DATASET**

```python
print(" AUTO-DETECTING ALL CLASSES FROM DATASET...")
print("=" * 60)

# ============================================
# STEP 1: Scan ALL XML files to find every unique class
# ============================================
all_class_names = set()
class_name_counts = Counter()

# Scan all weather folders
weather_types = ["Fog", "Rain", "Snow", "Sand"]
for weather in weather_types:
    weather_path = os.path.join(INPUT_PATH, weather, weather)
    
    # Find PASCAL_VOC folder
    pascal_folder = None
    if os.path.exists(weather_path):
        for item in os.listdir(weather_path):
            if "PASCAL_VOC" in item:
                pascal_folder = os.path.join(weather_path, item)
                break
    
    if pascal_folder is None:
        continue
    
    # Parse all XML files in this folder
    for xml_file in os.listdir(pascal_folder):
        if xml_file.endswith('.xml'):
            xml_path = os.path.join(pascal_folder, xml_file)
            try:
                tree = ET.parse(xml_path)
                root = tree.getroot()
                
                for obj in root.findall('object'):
                    name_elem = obj.find('name')
                    if name_elem is not None and name_elem.text:
                        class_name = name_elem.text.strip().lower()  # Normalize to lowercase
                        all_class_names.add(class_name)
                        class_name_counts[class_name] += 1
            except:
                continue

# Sort classes alphabetically for consistency
DETECTED_CLASSES = sorted(list(all_class_names))
num_classes = len(DETECTED_CLASSES)

print(f"\n\n\n DETECTED {num_classes} CLASSES:")
for idx, class_name in enumerate(DETECTED_CLASSES):
    print(f"   {idx}: {class_name} ({class_name_counts[class_name]} instances)")

print("\n" + "=" * 60)

# ============================================
# STEP 2: Convert XML to YOLO format with auto-detected classes
# ============================================
print(f"\n\n\n Converting annotations to YOLO format...")
print(f"   Using {num_classes} classes: {DETECTED_CLASSES}")

converted_count = 0
skipped_count = 0
class_distribution = Counter()

# Set random seed for reproducible train/val split
random.seed(42)

for img_path, xml_path in image_label_pairs:
    try:
        # Parse XML
        tree = ET.parse(xml_path)
        root = tree.getroot()
        
        # Get image dimensions
        size = root.find('size')
        width = int(size.find('width').text)
        height = int(size.find('height').text)
        
        # Convert objects to YOLO format
        yolo_lines = []
        for obj in root.findall('object'):
            # Get class name
            name_elem = obj.find('name')
            if name_elem is None or not name_elem.text:
                continue
            
            class_name = name_elem.text.strip().lower()  # Normalize
            
            # Get class ID from our detected classes list
            if class_name not in DETECTED_CLASSES:
                skipped_count += 1
                continue
            
            class_id = DETECTED_CLASSES.index(class_name)
            class_distribution[class_name] += 1
            
            # Get bounding box
            bbox = obj.find('bndbox')
            if bbox is None:
                continue
            
            xmin = float(bbox.find('xmin').text)
            ymin = float(bbox.find('ymin').text)
            xmax = float(bbox.find('xmax').text)
            ymax = float(bbox.find('ymax').text)
            
            # Convert to YOLO format (normalized center coordinates)
            x_center = ((xmin + xmax) / 2.0) / width
            y_center = ((ymin + ymax) / 2.0) / height
            box_width = (xmax - xmin) / width
            box_height = (ymax - ymin) / height
            
            # Clip to valid range [0, 1]
            x_center = max(0.0, min(1.0, x_center))
            y_center = max(0.0, min(1.0, y_center))
            box_width = max(0.0, min(1.0, box_width))
            box_height = max(0.0, min(1.0, box_height))
            
            # Add to YOLO lines
            yolo_lines.append(f"{class_id} {x_center:.6f} {y_center:.6f} {box_width:.6f} {box_height:.6f}")
        
        if not yolo_lines:
            skipped_count += 1
            continue
        
        # Random 80/20 train/val split
        is_train = random.random() < 0.8
        
        # Copy image to appropriate folder
        img_dest = IMAGES_TRAIN if is_train else IMAGES_VAL
        img_filename = os.path.basename(img_path)
        shutil.copy(img_path, os.path.join(img_dest, img_filename))
        
        # Save YOLO label
        label_dest = LABELS_TRAIN if is_train else LABELS_VAL
        label_filename = os.path.splitext(img_filename)[0] + '.txt'
        label_path = os.path.join(label_dest, label_filename)
        
        with open(label_path, 'w') as f:
            f.write('\n'.join(yolo_lines))
        
        converted_count += 1
        
    except Exception as e:
        skipped_count += 1
        continue

print(f"\n\n\n\n\n Conversion complete!")
print(f"   Converted: {converted_count} images")
print(f"   Skipped: {skipped_count} images")

print(f"\n\n\n\n Object distribution by class:")
for class_name in DETECTED_CLASSES:
    count = class_distribution[class_name]
    print(f"   {class_name}: {count} objects")

# ============================================
# STEP 3: Count train/val images
# ============================================
train_count = len([f for f in os.listdir(IMAGES_TRAIN) if f.endswith(('.jpg', '.png', '.jpeg'))])
val_count = len([f for f in os.listdir(IMAGES_VAL) if f.endswith(('.jpg', '.png', '.jpeg'))])

print(f"\n\n\n\n Dataset split:")
print(f"   Training: {train_count} images")
print(f"   Validation: {val_count} images")

# ============================================
# STEP 4: Create data.yaml with auto-detected classes
# ============================================
data_yaml_content = {
    'path': WORK_DIR,
    'train': 'images/train',
    'val': 'images/val',
    'nc': num_classes,  # Auto-detected number
    'names': DETECTED_CLASSES  # Auto-detected class names
}

yaml_path = f"{WORK_DIR}/data.yaml"
with open(yaml_path, 'w') as f:
    yaml.dump(data_yaml_content, f, default_flow_style=False, sort_keys=False)

print(f"\n\n\n\n\n data.yaml created successfully!")
print(f"\n\n\n\n\n Contents of data.yaml:")
print("-" * 60)
print(open(yaml_path).read())
print("-" * 60)

# ============================================
# STEP 5: Verify the conversion
# ============================================
print(f"\n\n\n\n VERIFICATION:")

# Check class IDs in generated labels
all_label_ids = set()
for label_dir in [LABELS_TRAIN, LABELS_VAL]:
    for label_file in os.listdir(label_dir):
        if label_file.endswith('.txt'):
            with open(os.path.join(label_dir, label_file), 'r') as f:
                for line in f:
                    parts = line.strip().split()
                    if len(parts) >= 5:
                        class_id = int(parts[0])
                        all_label_ids.add(class_id)

unique_ids = sorted(list(all_label_ids))
expected_ids = list(range(num_classes))

print(f"   Class IDs in labels: {unique_ids}")
print(f"   Expected range: {expected_ids}")

if unique_ids == expected_ids:
    print(f"    Perfect! All {num_classes} classes are properly mapped (0-{num_classes-1})")
else:
    print(f"     Warning: Class ID mismatch detected")
    missing = set(expected_ids) - set(unique_ids)
    extra = set(unique_ids) - set(expected_ids)
    if missing:
        print(f"   Missing IDs: {missing}")
    if extra:
        print(f"   Extra IDs: {extra}")

print("\n\n\n\n" + "=" * 60)
print(" AUTO-DETECTION AND CONVERSION COMPLETE!")
print("=" * 60)
```

# **14. Draw YOLO format bounding boxes on image.**

```python
def draw_yolo_boxes(image_path, label_path, class_names):
    """Draw YOLO format bounding boxes on image."""
    img = cv2.imread(image_path)
    if img is None:
        return None
    
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    h, w = img.shape[:2]
    
    if os.path.exists(label_path):
        with open(label_path) as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) < 5:
                    continue
                
                class_id = int(parts[0])
                x_center, y_center = float(parts[1]), float(parts[2])
                box_width, box_height = float(parts[3]), float(parts[4])
                
                # Convert to pixel coordinates
                x1 = int((x_center - box_width/2) * w)
                y1 = int((y_center - box_height/2) * h)
                x2 = int((x_center + box_width/2) * w)
                y2 = int((y_center + box_height/2) * h)
                
                # Draw box and label
                color = (0, 255, 0)
                cv2.rectangle(img, (x1, y1), (x2, y2), color, 2)
                
                label = class_names[class_id] if class_id < len(class_names) else str(class_id)
                cv2.putText(img, label, (x1, y1-5), cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 2)
    
    return img

# Display samples
train_images = [f for f in os.listdir(IMAGES_TRAIN) if f.endswith(('.jpg', '.png', '.jpeg'))]
sample_images = random.sample(train_images, min(6, len(train_images)))

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

for idx, img_name in enumerate(sample_images):
    img_path = os.path.join(IMAGES_TRAIN, img_name)
    label_path = os.path.join(LABELS_TRAIN, os.path.splitext(img_name)[0] + '.txt')
    
    vis = draw_yolo_boxes(img_path, label_path, DETECTED_CLASSES)
    if vis is not None:
        axes[idx].imshow(vis)
    axes[idx].set_title(img_name, fontsize=8)
    axes[idx].axis('off')

plt.suptitle('Training Samples with YOLO Labels', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

# **15. Initialize YOLOv8 nano model**

```python
# Initialize YOLOv8 nano model
model = YOLO('yolov8n.pt')

# Train the model
print(f"\n\n\n Starting training with {num_classes} classes...")
print(f"Classes: {DETECTED_CLASSES}\n")

results = model.train(
    data=f'{WORK_DIR}/data.yaml',
    epochs=20,
    imgsz=640,
    batch=-1,  # Auto batch size
    device=0,  # Use GPU
    patience=5,
    save=True,
    project='/kaggle/working',
    name='runs/detect/train'
)

print("\n\n\n Training complete!")
```

```python
# Load best model
best_model_path = '/kaggle/working/runs/detect/train/weights/best.pt'
model = YOLO(best_model_path)

# Validate on validation set
metrics = model.val()

print("\n\n\n\n Validation Metrics:")
print(f"  mAP50: {metrics.box.map50:.4f}")
print(f"  mAP50-95: {metrics.box.map:.4f}")
print(f"  Precision: {metrics.box.p.mean():.4f}")
print(f"  Recall: {metrics.box.r.mean():.4f}")
```

# **16. Running predictions on validation samples**

```python
# Get validation images
val_images = [f for f in os.listdir(IMAGES_VAL) if f.endswith(('.jpg', '.png', '.jpeg'))]
test_samples = random.sample(val_images, min(4, len(val_images)))

print("\n\n Running predictions on validation samples...\n")

for img_name in test_samples:
    img_path = os.path.join(IMAGES_VAL, img_name)
    
    # Run prediction
    results = model.predict(img_path, conf=0.25, device=0)
    
    # Display results
    result_img = results[0].plot()
    result_img = cv2.cvtColor(result_img, cv2.COLOR_BGR2RGB)
    
    plt.figure(figsize=(12, 8))
    plt.imshow(result_img)
    plt.title(f'Predictions: {img_name}', fontsize=12, fontweight='bold')
    plt.axis('off')
    plt.tight_layout()
    plt.show()
    
    # Print detection details
    print(f"Detections in {img_name}:")
    for box in results[0].boxes:
        class_id = int(box.cls[0])
        confidence = float(box.conf[0])
        class_name = DETECTED_CLASSES[class_id]
        print(f"  - {class_name} (confidence: {confidence:.2f})")
    print("-" * 50)
```
