## 1. Paths & Data Loading

```python
import os

INPUT_PATH = "/kaggle/input/datasets/orvile/dawn-detection-in-adverse-weather-nature/DAWN"

WEATHER_TYPES = ["Fog", "Rain", "Snow", "Sand"]

records = []

print("Scanning dataset...\n")

for weather in WEATHER_TYPES:

    # image folder
    img_dir = os.path.join(INPUT_PATH, weather, weather)

    # xml annotation folder
    xml_dir = os.path.join(INPUT_PATH, weather, f"{weather}_PASCAL_VOC")

    print(f"Weather: {weather}")
    print("Image dir:", img_dir)
    print("XML dir  :", xml_dir)

    if not os.path.exists(img_dir):
        print("[WARN] Image folder missing\n")
        continue

    if not os.path.exists(xml_dir):
        print("[WARN] XML folder missing\n")
        continue

    image_count = 0
    pair_count = 0

    for fname in os.listdir(img_dir):

        if not fname.lower().endswith((".jpg", ".jpeg", ".png")):
            continue

        image_count += 1

        img_path = os.path.join(img_dir, fname)

        # matching xml filename
        xml_name = os.path.splitext(fname)[0] + ".xml"

        xml_path = os.path.join(xml_dir, xml_name)

        if os.path.exists(xml_path):

            records.append({
                "img_path": img_path,
                "xml_path": xml_path,
                "weather": weather
            })

            pair_count += 1

    print(f"Images found : {image_count}")
    print(f"Valid pairs  : {pair_count}\n")

print("=" * 50)
print(f"Total image-annotation pairs: {len(records)}")

# sample
if len(records) > 0:
    print("\nSample Record:")
    print(records[0])
```

```python
import os

records = []

for weather in WEATHER_TYPES:

    # images
    img_dir = os.path.join(INPUT_PATH, weather, weather)

    # annotations
    xml_dir = os.path.join(
        INPUT_PATH,
        weather,
        weather,
        f"{weather}_PASCAL_VOC"
    )

    print(f"\n{weather}")
    print("IMG:", img_dir)
    print("XML:", xml_dir)

    if not os.path.exists(img_dir):
        print("Image folder missing")
        continue

    if not os.path.exists(xml_dir):
        print("XML folder missing")
        continue

    # create xml lookup
    xml_lookup = {}

    for x in os.listdir(xml_dir):

        if x.lower().endswith(".xml"):

            key = os.path.splitext(x)[0].lower()

            xml_lookup[key] = os.path.join(xml_dir, x)

    image_count = 0
    pair_count = 0

    for img_name in os.listdir(img_dir):

        if not img_name.lower().endswith((".jpg", ".jpeg", ".png")):
            continue

        image_count += 1

        key = os.path.splitext(img_name)[0].lower()

        if key in xml_lookup:

            records.append({
                "img_path": os.path.join(img_dir, img_name),
                "xml_path": xml_lookup[key],
                "weather": weather
            })

            pair_count += 1

    print("Images:", image_count)
    print("Pairs :", pair_count)

print("\nTOTAL RECORDS:", len(records))

if len(records) > 0:
    print("\nSample:")
    print(records[0])
```

```python
import os
import xml.etree.ElementTree as ET
import pandas as pd

def parse_annotation(xml_path):

    tree = ET.parse(xml_path)
    root = tree.getroot()

    # image size
    size = root.find("size")

    W = int(size.find("width").text) if size is not None else 1
    H = int(size.find("height").text) if size is not None else 1

    boxes = []

    # all objects
    for obj in root.findall(".//object"):

        name_tag = obj.find("name")
        bnd = obj.find("bndbox")

        if name_tag is None or bnd is None:
            continue

        try:
            xmin = float(bnd.find("xmin").text)
            ymin = float(bnd.find("ymin").text)
            xmax = float(bnd.find("xmax").text)
            ymax = float(bnd.find("ymax").text)

            boxes.append({
                "class": name_tag.text.strip(),
                "xmin": xmin,
                "ymin": ymin,
                "xmax": xmax,
                "ymax": ymax,
                "w": xmax - xmin,
                "h": ymax - ymin,
                "area": (xmax - xmin) * (ymax - ymin)
            })

        except:
            continue

    return {
        "img_w": W,
        "img_h": H,
        "boxes": boxes
    }


# build dataframe
rows = []

for rec in records:

    ann = parse_annotation(rec["xml_path"])

    for box in ann["boxes"]:

        rows.append({
            "weather": rec["weather"],
            "img_path": rec["img_path"],
            "img_w": ann["img_w"],
            "img_h": ann["img_h"],
            "class": box["class"],
            "bbox_w": box["w"],
            "bbox_h": box["h"],
            "bbox_area": box["area"]
        })

df_ann = pd.DataFrame(rows)

print("Rows collected:", len(df_ann))

# prevent crash if empty
if len(df_ann) > 0:

    df_ann["rel_w"] = df_ann["bbox_w"] / df_ann["img_w"]
    df_ann["rel_h"] = df_ann["bbox_h"] / df_ann["img_h"]
    df_ann["rel_area"] = (
        df_ann["bbox_area"] /
        (df_ann["img_w"] * df_ann["img_h"])
    )

    df_ann["aspect"] = (
        df_ann["bbox_w"] /
        (df_ann["bbox_h"] + 1e-6)
    )

    print(df_ann.shape)
    display(df_ann.head())

else:
    print("No annotations parsed from XML files.")
```

## 3. Preprocessing Pipeline

We build **two** pipelines:
1. **`train_transform`** — aggressive augmentation (flips, colour jitter, blur, coarse dropout)
2. **`val_transform`** — deterministic resize + normalise only

Both end with ImageNet normalisation and `ToTensorV2`.

```python
IMG_SIZE   = 128   # MLP baseline uses compact features; 128×128 is enough
IMAGENET_MEAN = (0.485, 0.456, 0.406)
IMAGENET_STD  = (0.229, 0.224, 0.225)

train_transform = A.Compose([
    A.Resize(IMG_SIZE, IMG_SIZE),
    A.HorizontalFlip(p=0.5),
    A.RandomBrightnessContrast(p=0.4),
    A.HueSaturationValue(hue_shift_limit=10, sat_shift_limit=20,
                         val_shift_limit=20, p=0.3),
    A.GaussianBlur(blur_limit=(3, 5), p=0.2),
    A.CoarseDropout(max_holes=8, max_height=16, max_width=16,
                    fill_value=0, p=0.2),
    A.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
    ToTensorV2(),
])

val_transform = A.Compose([
    A.Resize(IMG_SIZE, IMG_SIZE),
    A.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
    ToTensorV2(),
])

print("Train pipeline:")
print(train_transform)
print("\nVal pipeline:")
print(val_transform)
```

```python
# Visualise augmentation on one image
sample = random.choice(records)
orig   = cv2.cvtColor(cv2.imread(sample['img_path']), cv2.COLOR_BGR2RGB)
orig_r = cv2.resize(orig, (IMG_SIZE, IMG_SIZE))

aug_aug = A.Compose([
    A.Resize(IMG_SIZE, IMG_SIZE),
    A.HorizontalFlip(p=0.5),
    A.RandomBrightnessContrast(p=1.0),
    A.HueSaturationValue(p=1.0),
    A.GaussianBlur(p=1.0),
    A.CoarseDropout(p=1.0),
])

fig, axes = plt.subplots(1, 4, figsize=(14, 4))
axes[0].imshow(orig_r);   axes[0].set_title('Original');    axes[0].axis('off')
for i in range(1, 4):
    aug_img = aug_aug(image=orig)['image']
    axes[i].imshow(aug_img); axes[i].set_title(f'Augmented {i}'); axes[i].axis('off')

plt.suptitle('Preprocessing Pipeline — Augmentation Samples',
             fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig(f'{WORK_DIR}/pipeline_augmentation.png', dpi=100, bbox_inches='tight')
plt.show()
```

## 4. Feature Extraction & Dataset

For the **baseline MLP** we extract compact hand-crafted features per image:
- **RGB channel statistics** — mean + std for each channel (6 dims)
- **HSV statistics** — mean + std for H, S, V (6 dims)
- **Gradient energy** — mean Sobel magnitude (1 dim)
- **Laplacian variance** — blur metric (1 dim)
- **Histogram** — 8-bin histogram per RGB channel (24 dims)

Total: **38 features** — fast to compute, interpretable, good baseline.

```python
def extract_features(img_path: str, target_size: int = 128) -> np.ndarray:
    img = cv2.imread(img_path)
    if img is None:
        return None
    img = cv2.resize(img, (target_size, target_size))
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    gray    = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    feats = []

    # RGB channel stats
    for c in range(3):
        ch = img_rgb[:, :, c].astype(float) / 255.0
        feats += [ch.mean(), ch.std()]

    # HSV channel stats
    for c in range(3):
        ch = img_hsv[:, :, c].astype(float)
        ch /= (180. if c == 0 else 255.)
        feats += [ch.mean(), ch.std()]

    # Gradient energy
    sobelx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
    sobely = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
    feats.append(np.sqrt(sobelx**2 + sobely**2).mean() / 255.)

    # Laplacian variance (sharpness proxy)
    lap = cv2.Laplacian(gray, cv2.CV_64F).var()
    feats.append(np.log1p(lap) / 10.)   # rough normalisation

    # 8-bin colour histogram per RGB channel
    for c in range(3):
        hist, _ = np.histogram(img_rgb[:, :, c], bins=8, range=(0, 256))
        hist = hist.astype(float) / hist.sum()
        feats += hist.tolist()

    return np.array(feats, dtype=np.float32)

print(f"Feature vector size: {extract_features(records[0]['img_path']).shape[0]}")
```

```python
print("Extracting features (this may take ~1-2 min)...")

X_list, y_list = [], []
for rec in records:
    fv = extract_features(rec['img_path'])
    if fv is not None:
        X_list.append(fv)
        y_list.append(rec['weather'])

X_raw = np.stack(X_list)
le    = LabelEncoder()
y     = le.fit_transform(y_list)

print(f"Feature matrix: {X_raw.shape}")
print(f"Classes: {le.classes_}")

# Standard-scale
scaler  = StandardScaler()
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X_raw, y, test_size=0.2, random_state=SEED, stratify=y)
X_val_raw,   X_test_raw, y_val,   y_test  = train_test_split(
    X_test_raw, y_test, test_size=0.5, random_state=SEED, stratify=y_test)

X_train = scaler.fit_transform(X_train_raw)
X_val   = scaler.transform(X_val_raw)
X_test  = scaler.transform(X_test_raw)

print(f"Train/Val/Test split: {len(X_train)} / {len(X_val)} / {len(X_test)}")
```

```python
class FeatureDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.long)
    def __len__(self):  return len(self.y)
    def __getitem__(self, idx): return self.X[idx], self.y[idx]

def make_loaders(X_train, y_train, X_val, y_val, X_test, y_test, batch_size):
    train_dl = DataLoader(FeatureDataset(X_train, y_train),
                          batch_size=batch_size, shuffle=True)
    val_dl   = DataLoader(FeatureDataset(X_val,   y_val),
                          batch_size=batch_size)
    test_dl  = DataLoader(FeatureDataset(X_test,  y_test),
                          batch_size=batch_size)
    return train_dl, val_dl, test_dl
```

## 5. Baseline MLP Model

```python
class MLP(nn.Module):
    """Flexible MLP for weather classification."""
    def __init__(self,
                 input_dim: int,
                 hidden_dims: list,
                 num_classes: int,
                 dropout: float = 0.3,
                 use_bn: bool = True):
        super().__init__()
        layers = []
        prev = input_dim
        for h in hidden_dims:
            layers.append(nn.Linear(prev, h))
            if use_bn:
                layers.append(nn.BatchNorm1d(h))
            layers.append(nn.ReLU())
            layers.append(nn.Dropout(dropout))
            prev = h
        layers.append(nn.Linear(prev, num_classes))
        self.net = nn.Sequential(*layers)

    def forward(self, x):
        return self.net(x)

# Quick sanity check
dummy_model = MLP(input_dim=38, hidden_dims=[128, 64], num_classes=4)
dummy_out   = dummy_model(torch.randn(8, 38))
print("Output shape:", dummy_out.shape)
print(dummy_model)
```

---
## 6. Training Loop with MLflow Experiment Tracking

```python
def train_model(model, train_dl, val_dl, epochs, lr, weight_decay, device):
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.AdamW(model.parameters(), lr=lr,
                            weight_decay=weight_decay)
    scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs)

    history = {'train_loss': [], 'val_loss': [],
               'train_acc':  [], 'val_acc': []}
    best_val_acc = 0.0
    best_state   = None

    for epoch in range(1, epochs + 1):
        # ---- train ----
        model.train()
        t_loss, t_correct, t_total = 0., 0, 0
        for X_b, y_b in train_dl:
            X_b, y_b = X_b.to(device), y_b.to(device)
            optimizer.zero_grad()
            logits = model(X_b)
            loss   = criterion(logits, y_b)
            loss.backward()
            optimizer.step()
            t_loss    += loss.item() * len(y_b)
            t_correct += (logits.argmax(1) == y_b).sum().item()
            t_total   += len(y_b)

        # ---- val ----
        model.eval()
        v_loss, v_correct, v_total = 0., 0, 0
        with torch.no_grad():
            for X_b, y_b in val_dl:
                X_b, y_b = X_b.to(device), y_b.to(device)
                logits = model(X_b)
                loss   = criterion(logits, y_b)
                v_loss    += loss.item() * len(y_b)
                v_correct += (logits.argmax(1) == y_b).sum().item()
                v_total   += len(y_b)

        scheduler.step()

        tr_acc  = t_correct / t_total
        val_acc = v_correct / v_total
        tr_l    = t_loss    / t_total
        val_l   = v_loss    / v_total

        history['train_loss'].append(tr_l)
        history['val_loss'  ].append(val_l)
        history['train_acc' ].append(tr_acc)
        history['val_acc'   ].append(val_acc)

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_state   = {k: v.cpu().clone()
                            for k, v in model.state_dict().items()}

        # Log to MLflow each epoch
        mlflow.log_metrics({
            'train_loss': tr_l, 'val_loss': val_l,
            'train_acc':  tr_acc, 'val_acc': val_acc
        }, step=epoch)

        if epoch % 5 == 0 or epoch == 1:
            print(f"  Epoch {epoch:3d} | "
                  f"Train acc {tr_acc:.3f} loss {tr_l:.4f} | "
                  f"Val acc {val_acc:.3f} loss {val_l:.4f}")

    model.load_state_dict(best_state)
    return model, history, best_val_acc
```

```python
def evaluate_model(model, test_dl, device, class_names, run_name):
    model.eval()
    all_preds, all_labels = [], []
    with torch.no_grad():
        for X_b, y_b in test_dl:
            logits = model(X_b.to(device))
            all_preds  += logits.argmax(1).cpu().tolist()
            all_labels += y_b.tolist()

    acc = accuracy_score(all_labels, all_preds)
    print(f"\n{'='*50}")
    print(f"Test Accuracy: {acc:.4f}")
    print(classification_report(all_labels, all_preds, target_names=class_names))

    # Confusion matrix figure
    fig, ax = plt.subplots(figsize=(6, 5))
    cm = confusion_matrix(all_labels, all_preds)
    disp = ConfusionMatrixDisplay(cm, display_labels=class_names)
    disp.plot(cmap='Blues', ax=ax, colorbar=False)
    ax.set_title(f'Confusion Matrix — {run_name}', fontweight='bold')
    plt.tight_layout()
    cm_path = f'{WORK_DIR}/cm_{run_name}.png'
    plt.savefig(cm_path, dpi=120, bbox_inches='tight')
    plt.show()
    mlflow.log_artifact(cm_path)
    mlflow.log_metric('test_acc', acc)
    return acc, all_preds, all_labels
```

```python
def plot_history(history, run_name, save_dir):
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    epochs = range(1, len(history['train_loss']) + 1)

    axes[0].plot(epochs, history['train_loss'], label='Train')
    axes[0].plot(epochs, history['val_loss'],   label='Val')
    axes[0].set_title('Loss'); axes[0].set_xlabel('Epoch')
    axes[0].legend()

    axes[1].plot(epochs, history['train_acc'], label='Train')
    axes[1].plot(epochs, history['val_acc'],   label='Val')
    axes[1].set_title('Accuracy'); axes[1].set_xlabel('Epoch')
    axes[1].legend()

    plt.suptitle(f'Training Curves — {run_name}', fontsize=13, fontweight='bold')
    plt.tight_layout()
    path = f'{save_dir}/curves_{run_name}.png'
    plt.savefig(path, dpi=120, bbox_inches='tight')
    plt.show()
    mlflow.log_artifact(path)
```

## 7. Experiment Runs

```python
import os

MLFLOW_DIR = os.path.join(os.getcwd(), "mlruns")
```

```python
mlflow.set_tracking_uri(f'file://{MLFLOW_DIR}')
mlflow.set_experiment('DAWN_MLP_Weather_Classification')

NUM_CLASSES = len(le.classes_)
INPUT_DIM   = X_train.shape[1]

EXPERIMENTS = [
    {
        'run_name':    'MLP_shallow',
        'hidden_dims': [64],
        'dropout':     0.3,
        'use_bn':      True,
        'lr':          1e-3,
        'weight_decay':1e-4,
        'epochs':      40,
        'batch_size':  64,
    },
    {
        'run_name':    'MLP_medium',
        'hidden_dims': [256, 128, 64],
        'dropout':     0.3,
        'use_bn':      True,
        'lr':          1e-3,
        'weight_decay':1e-4,
        'epochs':      50,
        'batch_size':  64,
    },
    {
        'run_name':    'MLP_deep_lowdrop',
        'hidden_dims': [512, 256, 128, 64],
        'dropout':     0.15,
        'use_bn':      True,
        'lr':          5e-4,
        'weight_decay':1e-3,
        'epochs':      50,
        'batch_size':  32,
    },
    {
        'run_name':    'MLP_medium_no_bn',
        'hidden_dims': [256, 128, 64],
        'dropout':     0.4,
        'use_bn':      False,
        'lr':          1e-3,
        'weight_decay':1e-4,
        'epochs':      50,
        'batch_size':  64,
    },
]

run_results = []
```

```python
for cfg in EXPERIMENTS:
    print(f"\n{'='*55}")
    print(f"Run: {cfg['run_name']}")
    print(f"  hidden_dims={cfg['hidden_dims']}  dropout={cfg['dropout']}  "
          f"bn={cfg['use_bn']}  lr={cfg['lr']}  epochs={cfg['epochs']}")
    print('='*55)

    train_dl, val_dl, test_dl = make_loaders(
        X_train, y_train, X_val, y_val, X_test, y_test, cfg['batch_size'])

    model = MLP(
        input_dim   = INPUT_DIM,
        hidden_dims = cfg['hidden_dims'],
        num_classes = NUM_CLASSES,
        dropout     = cfg['dropout'],
        use_bn      = cfg['use_bn'],
    ).to(DEVICE)

    with mlflow.start_run(run_name=cfg['run_name']):
        # Log hyper-params
        mlflow.log_params({
            'hidden_dims':  str(cfg['hidden_dims']),
            'dropout':      cfg['dropout'],
            'use_bn':       cfg['use_bn'],
            'lr':           cfg['lr'],
            'weight_decay': cfg['weight_decay'],
            'epochs':       cfg['epochs'],
            'batch_size':   cfg['batch_size'],
            'input_dim':    INPUT_DIM,
            'num_classes':  NUM_CLASSES,
            'optimizer':    'AdamW',
            'scheduler':    'CosineAnnealingLR',
        })

        # Train
        model, history, best_val_acc = train_model(
            model, train_dl, val_dl,
            epochs       = cfg['epochs'],
            lr           = cfg['lr'],
            weight_decay = cfg['weight_decay'],
            device       = DEVICE,
        )

        # Plot & log training curves
        plot_history(history, cfg['run_name'], WORK_DIR)

        # Evaluate on test set
        test_acc, preds, labels = evaluate_model(
            model, test_dl, DEVICE, le.classes_, cfg['run_name'])

        # Save model weights
        model_path = f"{WORK_DIR}/model_{cfg['run_name']}.pt"
        torch.save(model.state_dict(), model_path)
        mlflow.log_artifact(model_path)

        run_results.append({
            'run_name':     cfg['run_name'],
            'hidden_dims':  str(cfg['hidden_dims']),
            'dropout':      cfg['dropout'],
            'use_bn':       cfg['use_bn'],
            'epochs':       cfg['epochs'],
            'best_val_acc': round(best_val_acc, 4),
            'test_acc':     round(test_acc, 4),
        })

print("\n All runs complete.")
```

## 8. Results Summary & Comparison

```python
results_df = pd.DataFrame(run_results).sort_values('test_acc', ascending=False)
print("\n" + "="*65)
print("EXPERIMENT RESULTS — sorted by test accuracy")
print("="*65)
print(results_df.to_string(index=False))
print("="*65)

best_run = results_df.iloc[0]
print(f"\n Best model: {best_run['run_name']}")
print(f"   Val Acc  : {best_run['best_val_acc']:.4f}")
print(f"   Test Acc : {best_run['test_acc']:.4f}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

x = np.arange(len(results_df))
axes[0].bar(x - 0.2, results_df['best_val_acc'], 0.4,
            label='Val Acc',  color='#4e79a7')
axes[0].bar(x + 0.2, results_df['test_acc'],     0.4,
            label='Test Acc', color='#e15759')
axes[0].set_xticks(x)
axes[0].set_xticklabels(results_df['run_name'], rotation=20, ha='right')
axes[0].set_ylabel('Accuracy')
axes[0].set_title('Val vs Test Accuracy by Run')
axes[0].legend()
axes[0].set_ylim(0, 1)
for xi, (v, t) in enumerate(zip(results_df['best_val_acc'], results_df['test_acc'])):
    axes[0].text(xi-0.2, v+0.01, f'{v:.3f}', ha='center', fontsize=8)
    axes[0].text(xi+0.2, t+0.01, f'{t:.3f}', ha='center', fontsize=8)

# Gap chart (over/underfitting)
gap = results_df['best_val_acc'] - results_df['test_acc']
colors = ['#59a14f' if g >= 0 else '#e15759' for g in gap]
axes[1].bar(x, gap, color=colors)
axes[1].set_xticks(x)
axes[1].set_xticklabels(results_df['run_name'], rotation=20, ha='right')
axes[1].axhline(0, color='black', linewidth=0.8)
axes[1].set_ylabel('Val Acc − Test Acc')
axes[1].set_title('Generalisation Gap')

plt.suptitle('Experiment Comparison', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(f'{WORK_DIR}/experiment_comparison.png', dpi=120, bbox_inches='tight')
plt.show()
```
