## 1. Paths & Data Loading
*Same dataset structure as Phase 2 — corrected to handle both XML folder layouts.*

````python
# ── Phase 2 paths (unchanged) ──────────────────────────────────────────────
INPUT_PATH = "/kaggle/input/datasets/orvile/dawn-detection-in-adverse-weather-nature/DAWN"
WEATHER_TYPES = ["Fog", "Rain", "Snow", "Sand"]
WEATHER_COLOR = {"Fog": "#4e79a7", "Rain": "#59a14f",
                 "Snow": "#edc948", "Sand": "#e15759"}

# ── Phase 3 working dirs ───────────────────────────────────────────────────
WORK_DIR   = "/kaggle/working/phase3"
MLFLOW_DIR = os.path.join(WORK_DIR, "mlruns")
CKPT_DIR   = os.path.join(WORK_DIR, "checkpoints")
for d in [WORK_DIR, MLFLOW_DIR, CKPT_DIR]:
    os.makedirs(d, exist_ok=True)

# ── Build records (mirrors Phase 2 corrected loader) ──────────────────────
records = []

for weather in WEATHER_TYPES:
    img_dir = os.path.join(INPUT_PATH, weather, weather)

    # Try both XML folder layouts seen in Phase 2
    xml_dir_a = os.path.join(INPUT_PATH, weather, f"{weather}_PASCAL_VOC")
    xml_dir_b = os.path.join(INPUT_PATH, weather, weather, f"{weather}_PASCAL_VOC")
    xml_dir   = xml_dir_a if os.path.exists(xml_dir_a) else xml_dir_b

    if not os.path.exists(img_dir):
        print(f"[WARN] {weather}: image folder missing — skipping")
        continue
    if not os.path.exists(xml_dir):
        print(f"[WARN] {weather}: XML folder missing — skipping")
        continue

    xml_lookup = {
        os.path.splitext(x)[0].lower(): os.path.join(xml_dir, x)
        for x in os.listdir(xml_dir) if x.lower().endswith('.xml')
    }

    n_img, n_pair = 0, 0
    for fname in os.listdir(img_dir):
        if not fname.lower().endswith(('.jpg', '.jpeg', '.png')):
            continue
        n_img += 1
        key = os.path.splitext(fname)[0].lower()
        if key in xml_lookup:
            records.append({'img_path': os.path.join(img_dir, fname),
                            'xml_path': xml_lookup[key],
                            'weather':  weather})
            n_pair += 1

    print(f"{weather:6s} | images: {n_img:4d}  pairs: {n_pair:4d}")

print(f"\nTotal records: {len(records)}")
````

## 2. Train / Val / Test Split
*70 / 15 / 15 stratified — consistent with Phase 2 label encoder.*

````python
# ── Label encoder (same as Phase 2 le) ────────────────────────────────────
le = LabelEncoder()
all_paths  = [r['img_path'] for r in records]
all_labels = le.fit_transform([r['weather'] for r in records])
NUM_CLASSES = len(le.classes_)
print(f"Classes ({NUM_CLASSES}): {le.classes_}")

# 70 / 15 / 15
X_tv,    X_test,  y_tv,    y_test  = train_test_split(
    all_paths, all_labels, test_size=0.15, random_state=SEED, stratify=all_labels)
X_train, X_val,   y_train, y_val   = train_test_split(
    X_tv, y_tv, test_size=0.176, random_state=SEED, stratify=y_tv)

print(f"Train : {len(X_train)}")
print(f"Val   : {len(X_val)}")
print(f"Test  : {len(X_test)}")

# ── Per-class weights (used by weighted CE loss & WeightedRandomSampler) ──
class_counts  = np.bincount(y_train, minlength=NUM_CLASSES).astype(float)
class_weights = 1.0 / (class_counts + 1e-6)
class_weights = class_weights / class_weights.sum() * NUM_CLASSES
CLASS_WEIGHT_TENSOR = torch.tensor(class_weights, dtype=torch.float32).to(DEVICE)

print("\nPer-class weights (weighted CE):")
for cls, w in zip(le.classes_, class_weights):
    print(f"  {cls:6s}: {w:.4f}  (n={int(class_counts[le.transform([cls])[0]])})")
````

## Requirement 1 — CNN Backbone (ResNet-18 / EfficientNet-B0) Fine-Tuned End-to-End

We replace the 38-dim hand-crafted feature extractor from Phase 2 with a pre-trained CNN loaded via `timm`. The classification head is always trainable; backbone layers are selectively unfrozen via `unfreeze_blocks`.

````python
IMAGENET_MEAN = (0.485, 0.456, 0.406)
IMAGENET_STD  = (0.229, 0.224, 0.225)
IMG_SIZE      = 224    # standard ImageNet resolution for pretrained backbones


def build_transforms(img_size: int, aug_strength: float = 1.0, mode: str = 'train'):
    """aug_strength ∈ [0,1] scales every augmentation probability — tuned by Optuna."""
    if mode == 'train':
        return A.Compose([
            A.Resize(img_size, img_size),
            A.HorizontalFlip(p=0.5),
            A.RandomBrightnessContrast(
                brightness_limit=0.2 * aug_strength,
                contrast_limit=0.2 * aug_strength, p=0.5),
            A.HueSaturationValue(
                hue_shift_limit=int(15 * aug_strength),
                sat_shift_limit=int(25 * aug_strength),
                val_shift_limit=int(25 * aug_strength), p=0.4),
            A.GaussianBlur(blur_limit=(3, 5), p=0.2 * aug_strength),
            A.GaussNoise(var_limit=(10.0, 50.0), p=0.2 * aug_strength),
            A.CoarseDropout(max_holes=6, max_height=24, max_width=24,
                            fill_value=0, p=0.2 * aug_strength),
            A.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
            ToTensorV2(),
        ])
    else:
        return A.Compose([
            A.Resize(img_size, img_size),
            A.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
            ToTensorV2(),
        ])


class DAWNImageDataset(Dataset):
    """Full-image dataset — replaces Phase 2 FeatureDataset."""
    def __init__(self, img_paths, labels, transform=None):
        self.img_paths = img_paths
        self.labels    = torch.tensor(labels, dtype=torch.long)
        self.transform = transform

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        img = cv2.imread(self.img_paths[idx])
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        if self.transform:
            img = self.transform(image=img)['image']
        return img, self.labels[idx]


def make_cnn_loaders(aug_strength, batch_size, use_sampler=False):
    tr_tf = build_transforms(IMG_SIZE, aug_strength, 'train')
    ev_tf = build_transforms(IMG_SIZE, aug_strength, 'eval')

    train_ds = DAWNImageDataset(X_train, y_train, tr_tf)
    val_ds   = DAWNImageDataset(X_val,   y_val,   ev_tf)
    test_ds  = DAWNImageDataset(X_test,  y_test,  ev_tf)

    if use_sampler:
        sample_w = class_weights[y_train]
        sampler  = WeightedRandomSampler(
            weights=torch.tensor(sample_w, dtype=torch.float64),
            num_samples=len(train_ds), replacement=True)
        train_dl = DataLoader(train_ds, batch_size=batch_size, sampler=sampler,
                              num_workers=2, pin_memory=True)
    else:
        train_dl = DataLoader(train_ds, batch_size=batch_size, shuffle=True,
                              num_workers=2, pin_memory=True)

    val_dl  = DataLoader(val_ds,  batch_size=batch_size, num_workers=2, pin_memory=True)
    test_dl = DataLoader(test_ds, batch_size=batch_size, num_workers=2, pin_memory=True)
    return train_dl, val_dl, test_dl


print("DAWNImageDataset and make_cnn_loaders defined.")
````

````python
class CNNClassifier(nn.Module):
    """
    Pre-trained CNN backbone (timm) + custom classification head.
    Backbone layers are selectively unfrozen from the top.
    """
    def __init__(self, backbone_name, num_classes, unfreeze_blocks, dropout):
        super().__init__()
        self.backbone_name = backbone_name

        # Load backbone, strip classifier
        self.backbone = timm.create_model(
            backbone_name, pretrained=True, num_classes=0, global_pool='avg')
        feat_dim = self.backbone.num_features

        # Freeze all backbone params first
        for p in self.backbone.parameters():
            p.requires_grad = False

        # Unfreeze top `unfreeze_blocks` quarters of backbone params
        if unfreeze_blocks > 0:
            params = list(self.backbone.parameters())
            n_unfreeze = max(1, len(params) * unfreeze_blocks // 4)
            for p in params[-n_unfreeze:]:
                p.requires_grad = True

        # Classification head — always trainable
        self.head = nn.Sequential(
            nn.Dropout(p=dropout),
            nn.Linear(feat_dim, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(),
            nn.Dropout(p=dropout * 0.5),
            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        feats = self.backbone(x)   # (B, feat_dim)
        return self.head(feats)    # (B, num_classes)

    def get_embedding(self, x):
        """Returns feature vector before head — used in YOLO fusion."""
        with torch.no_grad():
            return self.backbone(x)


def build_backbone(backbone_name, num_classes, unfreeze_blocks, dropout):
    return CNNClassifier(backbone_name, num_classes, unfreeze_blocks, dropout)


# ── Sanity check ───────────────────────────────────────────────────────────
for _name in ['resnet18', 'efficientnet_b0']:
    _m  = build_backbone(_name, NUM_CLASSES, unfreeze_blocks=1, dropout=0.3).to(DEVICE)
    _x  = torch.randn(2, 3, IMG_SIZE, IMG_SIZE).to(DEVICE)
    _out = _m(_x)
    n_trainable = sum(p.numel() for p in _m.parameters() if p.requires_grad)
    print(f"{_name:20s} | output: {_out.shape} | trainable params: {n_trainable:,}")
    del _m, _x, _out
````

## Requirement 2 — Hyperparameter Search (Optuna)

Optuna TPE sampler searches o
ver **8 hyperparameters** across 30 trials. Every trial is logged as a nested MLflow run inside the parent HPO run.

````python
def run_epoch(model, loader, criterion, optimizer, device, mode='train'):
    model.train() if mode == 'train' else model.eval()
    total_loss = correct = total = 0
    ctx = torch.enable_grad() if mode == 'train' else torch.no_grad()
    with ctx:
        for imgs, labels in loader:
            imgs, labels = imgs.to(device), labels.to(device)
            logits = model(imgs)
            loss   = criterion(logits, labels)
            if mode == 'train':
                optimizer.zero_grad()
                loss.backward()
                nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                optimizer.step()
            total_loss += loss.item() * len(labels)
            correct    += (logits.argmax(1) == labels).sum().item()
            total      += len(labels)
    return total_loss / total, correct / total


def train_cnn(model, train_dl, val_dl, cfg, device, mlflow_log=False):
    criterion = nn.CrossEntropyLoss(weight=CLASS_WEIGHT_TENSOR)
    optimizer = optim.AdamW(
        filter(lambda p: p.requires_grad, model.parameters()),
        lr=cfg['lr'], weight_decay=cfg['weight_decay'])
    scheduler = optim.lr_scheduler.CosineAnnealingLR(
        optimizer, T_max=cfg['epochs'], eta_min=cfg['lr'] * 0.01)

    history    = {'train_loss': [], 'val_loss': [], 'train_acc': [], 'val_acc': []}
    best_val   = 0.0
    best_state = None

    for epoch in range(1, cfg['epochs'] + 1):
        tr_loss, tr_acc = run_epoch(model, train_dl, criterion, optimizer, device, 'train')
        vl_loss, vl_acc = run_epoch(model, val_dl,   criterion, None,      device, 'eval')
        scheduler.step()

        history['train_loss'].append(tr_loss)
        history['val_loss'  ].append(vl_loss)
        history['train_acc' ].append(tr_acc)
        history['val_acc'   ].append(vl_acc)

        if vl_acc > best_val:
            best_val   = vl_acc
            best_state = copy.deepcopy(model.state_dict())

        if mlflow_log:
            mlflow.log_metrics({
                'train_loss': tr_loss, 'val_loss': vl_loss,
                'train_acc':  tr_acc,  'val_acc':  vl_acc,
                'lr': optimizer.param_groups[0]['lr']
            }, step=epoch)

        if epoch % 5 == 0 or epoch == 1:
            print(f"  Epoch {epoch:3d} | "
                  f"Train {tr_acc:.3f} ({tr_loss:.4f}) | "
                  f"Val   {vl_acc:.3f} ({vl_loss:.4f})")

    model.load_state_dict(best_state)
    return model, history, best_val


def evaluate_cnn(model, test_dl, device, class_names, run_name, save_dir):
    model.eval()
    preds, gts = [], []
    with torch.no_grad():
        for imgs, labels in test_dl:
            preds += model(imgs.to(device)).argmax(1).cpu().tolist()
            gts   += labels.tolist()

    acc = accuracy_score(gts, preds)
    f1  = f1_score(gts, preds, average='macro')

    print(f"\n{'='*50}")
    print(f"Test Accuracy : {acc:.4f}")
    print(f"Macro F1      : {f1:.4f}")
    print(classification_report(gts, preds, target_names=class_names))

    fig, ax = plt.subplots(figsize=(6, 5))
    cm   = confusion_matrix(gts, preds)
    disp = ConfusionMatrixDisplay(cm, display_labels=class_names)
    disp.plot(cmap='Blues', ax=ax, colorbar=False)
    ax.set_title(f'Confusion Matrix — {run_name}', fontweight='bold')
    plt.tight_layout()
    cm_path = f"{save_dir}/cm_{run_name}.png"
    plt.savefig(cm_path, dpi=120, bbox_inches='tight')
    plt.show()
    mlflow.log_artifact(cm_path)
    mlflow.log_metrics({'test_acc': acc, 'test_f1_macro': f1})
    return acc, f1


def plot_history(history, run_name, save_dir):
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    ep = range(1, len(history['train_loss']) + 1)
    axes[0].plot(ep, history['train_loss'], label='Train')
    axes[0].plot(ep, history['val_loss'],   label='Val')
    axes[0].set_title('Loss'); axes[0].legend()
    axes[1].plot(ep, history['train_acc'], label='Train')
    axes[1].plot(ep, history['val_acc'],   label='Val')
    axes[1].set_title('Accuracy'); axes[1].legend()
    plt.suptitle(f'Training Curves — {run_name}', fontsize=13, fontweight='bold')
    plt.tight_layout()
    path = f"{save_dir}/curves_{run_name}.png"
    plt.savefig(path, dpi=120, bbox_inches='tight')
    plt.show()
    mlflow.log_artifact(path)


print("Training utilities defined.")
````

````python
HPO_EPOCHS = 10    # short epochs per trial; increase for a longer search
N_TRIALS   = 30

mlflow.set_tracking_uri(f'file://{MLFLOW_DIR}')
mlflow.set_experiment('DAWN_Phase3_CNN')


def optuna_objective(trial):
    cfg = {
        'backbone':        trial.suggest_categorical(
                               'backbone', ['resnet18', 'efficientnet_b0']),
        'lr':              trial.suggest_float('lr',          1e-5, 1e-2, log=True),
        'weight_decay':    trial.suggest_float('weight_decay', 1e-5, 1e-2, log=True),
        'dropout':         trial.suggest_float('dropout',      0.1,  0.6),
        'unfreeze_blocks': trial.suggest_int  ('unfreeze_blocks', 0, 4),
        'aug_strength':    trial.suggest_float('aug_strength', 0.3,  1.0),
        'batch_size':      trial.suggest_categorical('batch_size', [16, 32, 64]),
        'use_sampler':     trial.suggest_categorical('use_sampler', [True, False]),
        'epochs':          HPO_EPOCHS,
    }

    with mlflow.start_run(run_name=f'trial_{trial.number:02d}', nested=True):
        mlflow.log_params(cfg)

        train_dl, val_dl, _ = make_cnn_loaders(
            cfg['aug_strength'], cfg['batch_size'], cfg['use_sampler'])

        model = build_backbone(
            cfg['backbone'], NUM_CLASSES,
            cfg['unfreeze_blocks'], cfg['dropout']
        ).to(DEVICE)

        _, _, best_val_acc = train_cnn(
            model, train_dl, val_dl, cfg, DEVICE, mlflow_log=True)

        mlflow.log_metric('best_val_acc', best_val_acc)

    return best_val_acc


print(f"Optuna search: {N_TRIALS} trials × {HPO_EPOCHS} epochs each.")
print("Starting...")

with mlflow.start_run(run_name='HPO_parent'):
    study = optuna.create_study(
        direction='maximize',
        sampler=TPESampler(seed=SEED),
        study_name='dawn_cnn_hpo')
    study.optimize(optuna_objective, n_trials=N_TRIALS, show_progress_bar=True)

    best = study.best_trial
    print(f"\n✅ Best trial #{best.number}")
    print(f"   Val Acc : {best.value:.4f}")
    print(f"   Params  : {best.params}")
    mlflow.log_metric('best_hpo_val_acc', best.value)
    mlflow.log_params({f'best_{k}': v for k, v in best.params.items()})

BEST_CFG = best.params
print("\nBest config stored in BEST_CFG.")
````

````python
trials_df = study.trials_dataframe().rename(columns={'value': 'val_acc'})

fig, axes = plt.subplots(1, 3, figsize=(16, 4))

# Optimisation history
trials_df['best_so_far'] = trials_df['val_acc'].cummax()
axes[0].plot(trials_df.index, trials_df['val_acc'],
             'o-', alpha=0.5, ms=4, label='Trial val acc')
axes[0].plot(trials_df.index, trials_df['best_so_far'],
             color='red', lw=2, label='Best so far')
axes[0].set_title('HPO Optimisation History')
axes[0].set_xlabel('Trial'); axes[0].legend()

# Backbone vs accuracy
bb_col = 'params_backbone'
if bb_col in trials_df.columns:
    sns.boxplot(data=trials_df, x=bb_col, y='val_acc',
                palette='Set2', ax=axes[1])
    axes[1].set_title('Val Acc by Backbone'); axes[1].set_xlabel('')

# LR vs accuracy
lr_col = 'params_lr'
if lr_col in trials_df.columns:
    axes[2].scatter(np.log10(trials_df[lr_col]), trials_df['val_acc'],
                    alpha=0.6, s=30, c='#4e79a7')
    axes[2].set_title('log10(LR) vs Val Acc')
    axes[2].set_xlabel('log10(LR)'); axes[2].set_ylabel('Val Acc')

plt.suptitle('Optuna HPO Analysis', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(f'{WORK_DIR}/hpo_analysis.png', dpi=120, bbox_inches='tight')
plt.show()

print("Top 5 trials:")
cols = ['number', 'val_acc', 'params_backbone', 'params_lr', 'params_dropout']
print(trials_df[[c for c in cols if c in trials_df.columns]]
      .nlargest(5, 'val_acc').to_string(index=False))
````

## Requirement 3 — Class-Imbalance Handling (Weighted Loss / Oversampling)

Three strategies are compared head-to-head using the best backbone from HPO:

| Strategy | Mechanism |
|---|---|
| `no_correction` | Standard CE loss, uniform sampling |
| `weighted_loss` | Per-class weighted CE loss |
| `weighted_sampler` | `WeightedRandomSampler` (oversampling minority) |

````python
ABLATION_EPOCHS = 15

IMBALANCE_STRATEGIES = [
    {'name': 'no_correction',    'weighted_loss': False, 'use_sampler': False},
    {'name': 'weighted_loss',    'weighted_loss': True,  'use_sampler': False},
    {'name': 'weighted_sampler', 'weighted_loss': False, 'use_sampler': True},
]

imbalance_results = []
mlflow.set_experiment('DAWN_Phase3_Imbalance_Ablation')

for strat in IMBALANCE_STRATEGIES:
    print(f"\n{'='*55}")
    print(f"Strategy: {strat['name']}")
    print('='*55)

    train_dl, val_dl, test_dl = make_cnn_loaders(
        aug_strength=BEST_CFG.get('aug_strength', 0.7),
        batch_size=32,
        use_sampler=strat['use_sampler'])

    model = build_backbone(
        BEST_CFG['backbone'], NUM_CLASSES,
        BEST_CFG.get('unfreeze_blocks', 2),
        BEST_CFG.get('dropout', 0.3)
    ).to(DEVICE)

    # Criterion: weighted or standard CE
    weight_t  = CLASS_WEIGHT_TENSOR if strat['weighted_loss'] else None
    criterion = nn.CrossEntropyLoss(weight=weight_t)

    optimizer = optim.AdamW(
        filter(lambda p: p.requires_grad, model.parameters()),
        lr=BEST_CFG.get('lr', 1e-4),
        weight_decay=BEST_CFG.get('weight_decay', 1e-4))
    scheduler = optim.lr_scheduler.CosineAnnealingLR(
        optimizer, T_max=ABLATION_EPOCHS)

    best_val, best_st = 0., None
    with mlflow.start_run(run_name=strat['name']):
        mlflow.log_params({**BEST_CFG, **strat, 'epochs': ABLATION_EPOCHS})

        for epoch in range(1, ABLATION_EPOCHS + 1):
            tl, ta = run_epoch(model, train_dl, criterion, optimizer, DEVICE, 'train')
            vl, va = run_epoch(model, val_dl,   criterion, None,      DEVICE, 'eval')
            scheduler.step()
            if va > best_val:
                best_val = va
                best_st  = copy.deepcopy(model.state_dict())
            mlflow.log_metrics({'train_acc': ta, 'val_acc': va,
                                'train_loss': tl, 'val_loss': vl}, step=epoch)
            if epoch % 5 == 0:
                print(f"  Epoch {epoch:3d} | Train {ta:.3f} | Val {va:.3f}")

        model.load_state_dict(best_st)

        # Test evaluation
        model.eval()
        preds, gts = [], []
        with torch.no_grad():
            for imgs, labels in test_dl:
                preds += model(imgs.to(DEVICE)).argmax(1).cpu().tolist()
                gts   += labels.tolist()

        acc = accuracy_score(gts, preds)
        f1  = f1_score(gts, preds, average='macro')
        mlflow.log_metrics({'test_acc': acc, 'test_f1_macro': f1})

        print(f"  → Test Acc: {acc:.4f}  Macro F1: {f1:.4f}")

    imbalance_results.append({'strategy':  strat['name'],
                              'best_val':  round(best_val, 4),
                              'test_acc':  round(acc, 4),
                              'test_f1':   round(f1, 4)})

imbalance_df = pd.DataFrame(imbalance_results)
print("\n" + "="*55)
print("Class-Imbalance Ablation Results")
print("="*55)
print(imbalance_df.to_string(index=False))
````

## Requirement 4 — YOLO Feature Fusion


YOLOv8 detections are used as **extra input features** alongside the CNN embedding:

```
Image ──► CNN backbone ──► embedding (feat_dim)
                                          │
                                       concat ──► FusionHead ──► class logits
                                          │
Image ──► YOLOv8 ──► detection stats ──► yolo_feats (11-dim)
```

**YOLO feature vector (11 dims per image)**:
- Total detections count (1)
- Per-class detection count: car, person, truck, bus, motorcycle, bicycle (6)
- Mean confidence, max confidence, std confidence (3)
- Detection density = detections / image area (1)

````python
from ultralytics import YOLO as UltralyticsYOLO

YOLO_MODEL_PATH = '/kaggle/working/phase1/runs/detect/train/weights/best.pt'

# Fallback: use pretrained YOLOv8n if Phase 1 weights not found
if not os.path.exists(YOLO_MODEL_PATH):
    print("[WARN] Phase 1 YOLO weights not found — using pretrained YOLOv8n as fallback")
    YOLO_MODEL_PATH = 'yolov8n.pt'

yolo_model = UltralyticsYOLO(YOLO_MODEL_PATH)
yolo_model.fuse()

# YOLO class names for feature extraction
YOLO_TRACK_CLASSES = ['car', 'person', 'truck', 'bus', 'motorcycle', 'bicycle']
YOLO_FEAT_DIM      = 1 + len(YOLO_TRACK_CLASSES) + 3 + 1   # = 11


def extract_yolo_features(img_path: str, yolo_model, conf_thresh=0.25) -> np.ndarray:
    """Run YOLOv8 on one image and return an 11-dim feature vector."""
    results = yolo_model.predict(
        img_path, conf=conf_thresh, verbose=False, imgsz=640)[0]

    boxes  = results.boxes
    names  = results.names    # dict {id: class_name}

    feats = np.zeros(YOLO_FEAT_DIM, dtype=np.float32)

    if boxes is None or len(boxes) == 0:
        return feats

    confs      = boxes.conf.cpu().numpy()
    class_ids  = boxes.cls.cpu().numpy().astype(int)
    class_strs = [names.get(i, '').lower() for i in class_ids]

    n_det = len(confs)
    feats[0] = n_det

    # Per-class counts
    for ci, cls_name in enumerate(YOLO_TRACK_CLASSES):
        feats[1 + ci] = sum(1 for c in class_strs if c == cls_name)

    # Confidence stats
    feats[-4] = float(confs.mean())
    feats[-3] = float(confs.max())
    feats[-2] = float(confs.std() if len(confs) > 1 else 0.0)

    # Detection density (approx — normalise by 640²)
    feats[-1] = n_det / (640 * 640)

    return feats


# Test on one image
sample_fv = extract_yolo_features(records[0]['img_path'], yolo_model)
print(f"YOLO feature vector ({YOLO_FEAT_DIM} dims): {sample_fv}")
````

````python
YOLO_CACHE = os.path.join(WORK_DIR, 'yolo_features.npy')
YOLO_PATHS_CACHE = os.path.join(WORK_DIR, 'yolo_paths.json')

if os.path.exists(YOLO_CACHE):
    print("Loading cached YOLO features...")
    yolo_feats_all = np.load(YOLO_CACHE)
    with open(YOLO_PATHS_CACHE) as f:
        cached_paths = json.load(f)
else:
    print(f"Extracting YOLO features for {len(records)} images (may take several minutes)...")
    yolo_feats_all = []
    cached_paths   = []
    for i, rec in enumerate(records):
        fv = extract_yolo_features(rec['img_path'], yolo_model)
        yolo_feats_all.append(fv)
        cached_paths.append(rec['img_path'])
        if (i + 1) % 100 == 0:
            print(f"  {i+1}/{len(records)}")
    yolo_feats_all = np.stack(yolo_feats_all)
    np.save(YOLO_CACHE, yolo_feats_all)
    with open(YOLO_PATHS_CACHE, 'w') as f:
        json.dump(cached_paths, f)
    print("YOLO features cached.")

# Align YOLO features to train/val/test splits
path_to_idx = {p: i for i, p in enumerate(cached_paths)}

def get_yolo_split(paths):
    idxs = [path_to_idx[p] for p in paths if p in path_to_idx]
    return yolo_feats_all[idxs]

Y_train_yolo = get_yolo_split(X_train)
Y_val_yolo   = get_yolo_split(X_val)
Y_test_yolo  = get_yolo_split(X_test)

# Scale YOLO features
yolo_scaler   = StandardScaler()
Y_train_yolo  = yolo_scaler.fit_transform(Y_train_yolo)
Y_val_yolo    = yolo_scaler.transform(Y_val_yolo)
Y_test_yolo   = yolo_scaler.transform(Y_test_yolo)

print(f"YOLO feature matrix: {yolo_feats_all.shape}")
print(f"Train split: {Y_train_yolo.shape}")
````

````python
class FusionDataset(Dataset):
    """Yields (image_tensor, yolo_feature_vector, label) for the fusion model."""
    def __init__(self, img_paths, yolo_feats, labels, transform=None):
        self.img_paths  = img_paths
        self.yolo_feats = torch.tensor(yolo_feats, dtype=torch.float32)
        self.labels     = torch.tensor(labels,     dtype=torch.long)
        self.transform  = transform

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        img = cv2.imread(self.img_paths[idx])
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        if self.transform:
            img = self.transform(image=img)['image']
        return img, self.yolo_feats[idx], self.labels[idx]


def make_fusion_loaders(aug_strength, batch_size):
    tr_tf = build_transforms(IMG_SIZE, aug_strength, 'train')
    ev_tf = build_transforms(IMG_SIZE, aug_strength, 'eval')

    train_dl = DataLoader(
        FusionDataset(X_train, Y_train_yolo, y_train, tr_tf),
        batch_size=batch_size, shuffle=True, num_workers=2, pin_memory=True)
    val_dl   = DataLoader(
        FusionDataset(X_val,   Y_val_yolo,   y_val,   ev_tf),
        batch_size=batch_size, num_workers=2, pin_memory=True)
    test_dl  = DataLoader(
        FusionDataset(X_test,  Y_test_yolo,  y_test,  ev_tf),
        batch_size=batch_size, num_workers=2, pin_memory=True)
    return train_dl, val_dl, test_dl


print("FusionDataset defined.")
````

````python
class CNNYOLOFusionModel(nn.Module):
    """
    CNN embedding + YOLO detection features ──► weather class.

    Architecture:
        Image  ──► frozen/unfrozen CNN backbone ──► (feat_dim,)
        YOLO   ──► MLP encoder                  ──► (32,)
        concat ──► fusion MLP                   ──► (num_classes,)
    """
    def __init__(self, backbone_name, num_classes,
                 unfreeze_blocks, dropout, yolo_feat_dim):
        super().__init__()

        # CNN backbone (same factory as Requirement 1)
        self.cnn = timm.create_model(
            backbone_name, pretrained=True, num_classes=0, global_pool='avg')
        feat_dim = self.cnn.num_features

        for p in self.cnn.parameters():
            p.requires_grad = False
        if unfreeze_blocks > 0:
            params = list(self.cnn.parameters())
            n = max(1, len(params) * unfreeze_blocks // 4)
            for p in params[-n:]:
                p.requires_grad = True

        # YOLO encoder: projects 11-dim vector to 32-dim
        self.yolo_encoder = nn.Sequential(
            nn.Linear(yolo_feat_dim, 32),
            nn.BatchNorm1d(32),
            nn.ReLU(),
            nn.Dropout(dropout * 0.5)
        )

        # Fusion head
        fused_dim = feat_dim + 32
        self.fusion_head = nn.Sequential(
            nn.Dropout(dropout),
            nn.Linear(fused_dim, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(),
            nn.Dropout(dropout * 0.5),
            nn.Linear(256, num_classes)
        )

    def forward(self, img, yolo_feat):
        cnn_emb   = self.cnn(img)                      # (B, feat_dim)
        yolo_emb  = self.yolo_encoder(yolo_feat)       # (B, 32)
        fused     = torch.cat([cnn_emb, yolo_emb], 1)  # (B, feat_dim+32)
        return self.fusion_head(fused)                  # (B, num_classes)


# Sanity check
_fmodel = CNNYOLOFusionModel(
    BEST_CFG['backbone'], NUM_CLASSES,
    BEST_CFG.get('unfreeze_blocks', 2),
    BEST_CFG.get('dropout', 0.3),
    YOLO_FEAT_DIM
).to(DEVICE)
_img  = torch.randn(2, 3, IMG_SIZE, IMG_SIZE).to(DEVICE)
_yolo = torch.randn(2, YOLO_FEAT_DIM).to(DEVICE)
print("Fusion model output:", _fmodel(_img, _yolo).shape)
del _fmodel, _img, _yolo
````

````python
FUSION_EPOCHS = 25
FUSION_LR     = BEST_CFG.get('lr', 1e-4)
FUSION_BS     = BEST_CFG.get('batch_size', 32)

fusion_model = CNNYOLOFusionModel(
    BEST_CFG['backbone'], NUM_CLASSES,
    BEST_CFG.get('unfreeze_blocks', 2),
    BEST_CFG.get('dropout', 0.3),
    YOLO_FEAT_DIM
).to(DEVICE)

criterion = nn.CrossEntropyLoss(weight=CLASS_WEIGHT_TENSOR)
optimizer = optim.AdamW(
    filter(lambda p: p.requires_grad, fusion_model.parameters()),
    lr=FUSION_LR, weight_decay=BEST_CFG.get('weight_decay', 1e-4))
scheduler = optim.lr_scheduler.CosineAnnealingLR(
    optimizer, T_max=FUSION_EPOCHS, eta_min=FUSION_LR * 0.01)

train_dl, val_dl, test_dl = make_fusion_loaders(
    BEST_CFG.get('aug_strength', 0.7), FUSION_BS)

fusion_history = {'train_loss': [], 'val_loss': [], 'train_acc': [], 'val_acc': []}
best_fusion_val  = 0.
best_fusion_state = None

mlflow.set_experiment('DAWN_Phase3_CNN_YOLO_Fusion')

with mlflow.start_run(run_name='CNN_YOLO_Fusion'):
    mlflow.log_params({
        **BEST_CFG,
        'yolo_feat_dim':   YOLO_FEAT_DIM,
        'fusion_epochs':   FUSION_EPOCHS,
        'imbalance':       'weighted_CE_loss',
        'yolo_model_path': YOLO_MODEL_PATH,
    })
    mlflow.set_tags({'phase': '3', 'fusion': 'CNN+YOLO'})

    print(f"Training CNN+YOLO Fusion model — {FUSION_EPOCHS} epochs...")

    for epoch in range(1, FUSION_EPOCHS + 1):
        # ── Train ──────────────────────────────────────────────────────────
        fusion_model.train()
        t_loss = t_correct = t_total = 0
        for imgs, yfeats, labels in train_dl:
            imgs, yfeats, labels = imgs.to(DEVICE), yfeats.to(DEVICE), labels.to(DEVICE)
            optimizer.zero_grad()
            logits = fusion_model(imgs, yfeats)
            loss   = criterion(logits, labels)
            loss.backward()
            nn.utils.clip_grad_norm_(fusion_model.parameters(), 1.0)
            optimizer.step()
            t_loss    += loss.item() * len(labels)
            t_correct += (logits.argmax(1) == labels).sum().item()
            t_total   += len(labels)

        # ── Val ────────────────────────────────────────────────────────────
        fusion_model.eval()
        v_loss = v_correct = v_total = 0
        with torch.no_grad():
            for imgs, yfeats, labels in val_dl:
                imgs, yfeats, labels = imgs.to(DEVICE), yfeats.to(DEVICE), labels.to(DEVICE)
                logits = fusion_model(imgs, yfeats)
                loss   = criterion(logits, labels)
                v_loss    += loss.item() * len(labels)
                v_correct += (logits.argmax(1) == labels).sum().item()
                v_total   += len(labels)

        scheduler.step()

        tr_acc = t_correct / t_total; tr_l = t_loss / t_total
        vl_acc = v_correct / v_total; vl_l = v_loss / v_total

        fusion_history['train_loss'].append(tr_l)
        fusion_history['val_loss'  ].append(vl_l)
        fusion_history['train_acc' ].append(tr_acc)
        fusion_history['val_acc'   ].append(vl_acc)

        if vl_acc > best_fusion_val:
            best_fusion_val   = vl_acc
            best_fusion_state = copy.deepcopy(fusion_model.state_dict())

        mlflow.log_metrics({'train_loss': tr_l, 'val_loss': vl_l,
                            'train_acc':  tr_acc, 'val_acc': vl_acc}, step=epoch)

        if epoch % 5 == 0 or epoch == 1:
            print(f"  Epoch {epoch:3d} | Train {tr_acc:.3f} ({tr_l:.4f}) "
                  f"| Val {vl_acc:.3f} ({vl_l:.4f})")

    fusion_model.load_state_dict(best_fusion_state)

    # ── Plot curves ────────────────────────────────────────────────────────
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    ep = range(1, FUSION_EPOCHS + 1)
    axes[0].plot(ep, fusion_history['train_loss'], label='Train')
    axes[0].plot(ep, fusion_history['val_loss'],   label='Val')
    axes[0].set_title('Fusion Loss'); axes[0].legend()
    axes[1].plot(ep, fusion_history['train_acc'], label='Train')
    axes[1].plot(ep, fusion_history['val_acc'],   label='Val')
    axes[1].set_title('Fusion Accuracy'); axes[1].legend()
    plt.suptitle('CNN + YOLO Fusion — Training Curves', fontsize=13, fontweight='bold')
    plt.tight_layout()
    curves_path = f'{WORK_DIR}/curves_fusion.png'
    plt.savefig(curves_path, dpi=120, bbox_inches='tight')
    plt.show()
    mlflow.log_artifact(curves_path)

    # ── Test evaluation ────────────────────────────────────────────────────
    fusion_model.eval()
    f_preds, f_gts = [], []
    with torch.no_grad():
        for imgs, yfeats, labels in test_dl:
            logits = fusion_model(imgs.to(DEVICE), yfeats.to(DEVICE))
            f_preds += logits.argmax(1).cpu().tolist()
            f_gts   += labels.tolist()

    fusion_test_acc = accuracy_score(f_gts, f_preds)
    fusion_test_f1  = f1_score(f_gts, f_preds, average='macro')

    print(f"\n{'='*50}")
    print(f"CNN+YOLO Fusion — Test Accuracy : {fusion_test_acc:.4f}")
    print(f"CNN+YOLO Fusion — Macro F1      : {fusion_test_f1:.4f}")
    print(classification_report(f_gts, f_preds, target_names=le.classes_))

    fig, ax = plt.subplots(figsize=(6, 5))
    cm = confusion_matrix(f_gts, f_preds)
    ConfusionMatrixDisplay(cm, display_labels=le.classes_).plot(
        cmap='Blues', ax=ax, colorbar=False)
    ax.set_title('Confusion Matrix — CNN+YOLO Fusion', fontweight='bold')
    plt.tight_layout()
    cm_path = f'{WORK_DIR}/cm_fusion.png'
    plt.savefig(cm_path, dpi=120, bbox_inches='tight')
    plt.show()
    mlflow.log_artifact(cm_path)
    mlflow.log_metrics({'test_acc': fusion_test_acc, 'test_f1_macro': fusion_test_f1})

    # Save fusion checkpoint
    ckpt = f'{CKPT_DIR}/fusion_model_phase3.pt'
    torch.save({'model_state': fusion_model.state_dict(),
                'best_cfg':   BEST_CFG,
                'classes':    list(le.classes_),
                'test_acc':   fusion_test_acc,
                'test_f1':    fusion_test_f1}, ckpt)
    mlflow.log_artifact(ckpt)
    print(f"\n\n Fusion model saved: {ckpt}")
````
